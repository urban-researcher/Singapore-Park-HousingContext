```python
import geopandas as gpd
import pandas as pd 
import numpy as np

import os
# Working directory is explicitly configured in the first cell.
```

```python
parks = ["amk", "jurong", "punggol"]
ttl_df = pd.DataFrame(columns=[
    "park",
    "mean_greenness",
    "mean_openness",
    "mean_furniture",
    "median_greenness", 
    "median_openness",
    "median_furniture",
    "std_greenness",
    "std_openness",
    "std_furniture",
])
for park in parks:
    print(f"Processing {park}")
    in_fp = f"gis_files/hexagons/segmented/{park}_hex_agg.geojson"
    gdf = gpd.read_file(in_fp)
    # if some rows has "greenness_mean", "openness_mean", "furniture_mean" all Nah, then drop these rows
    gdf = gdf.dropna(subset=['greenness_mean', 'openness_mean', 'furniture_mean'], how='all')
    mean_greenness = gdf['greenness_mean'].mean()
    mean_openness = gdf['openness_mean'].mean()
    mean_furniture = gdf['furniture_mean'].mean()
    median_greenness = gdf['greenness_mean'].median()
    median_openness = gdf['openness_mean'].median()
    median_furniture = gdf['furniture_mean'].median()
    std_greenness = gdf['greenness_mean'].std()
    std_openness = gdf['openness_mean'].std()
    std_furniture = gdf['furniture_mean'].std()
    ttl_df = pd.merge(ttl_df, pd.DataFrame({
        "park": [park],
        "mean_greenness": [mean_greenness],
        "mean_openness": [mean_openness],
        "mean_furniture": [mean_furniture],
        "median_greenness": [median_greenness],
        "median_openness": [median_openness],
        "median_furniture": [median_furniture],
        "std_greenness": [std_greenness],
        "std_openness": [std_openness],
        "std_furniture": [std_furniture],
    }), how='outer') 

ttl_df.to_csv('analysis/generate_stas/parks_greenness_openness_furniture_summary.csv', index=False)
```

```python
parks = ["amk", "jurong", "punggol"]

ttl_df = pd.DataFrame(columns=[
    "park",
    "mean_safety",
    "mean_lively",
    "mean_wealthy",
    "mean_beautiful",
    "mean_boring",
    "mean_depressing",
    "median_safety",
    "median_lively",
    "median_wealthy",
    "median_beautiful",
    "median_boring",
    "median_depressing",
    "std_safety",
    "std_lively",
    "std_wealthy",
    "std_beautiful",
    "std_boring",
    "std_depressing",
])
for park in parks:
    print(f"Processing {park}")
    in_fp = f"gis_files/hexagons/human_perception/{park}_hex_agg.geojson"
    gdf = gpd.read_file(in_fp)
    # if some rows has "greenness_mean", "openness_mean", "furniture_mean" all Nah, then drop these rows
    gdf = gdf.dropna(subset=["safety_score_mean", "lively_score_mean", "wealthy_score_mean", "beautiful_score_mean", "boring_score_mean", "depressing_score_mean"], how='all')
    mean_safety = gdf['safety_score_mean'].mean()
    mean_lively = gdf['lively_score_mean'].mean()
    mean_wealthy = gdf['wealthy_score_mean'].mean()
    mean_beautiful = gdf['beautiful_score_mean'].mean()
    mean_boring = gdf['boring_score_mean'].mean()
    mean_depressing = gdf['depressing_score_mean'].mean()
    median_safety = gdf['safety_score_mean'].median()
    median_lively = gdf['lively_score_mean'].median()
    median_wealthy = gdf['wealthy_score_mean'].median()
    median_beautiful = gdf['beautiful_score_mean'].median()
    median_boring = gdf['boring_score_mean'].median()
    median_depressing = gdf['depressing_score_mean'].median()
    std_safety = gdf['safety_score_mean'].std()
    std_lively = gdf['lively_score_mean'].std()
    std_wealthy = gdf['wealthy_score_mean'].std()
    std_beautiful = gdf['beautiful_score_mean'].std()
    std_boring = gdf['boring_score_mean'].std()
    std_depressing = gdf['depressing_score_mean'].std()
    ttl_df = pd.merge(ttl_df, pd.DataFrame({
        "park": [park],
        "mean_safety": [mean_safety],
        "mean_lively": [mean_lively],
        "mean_wealthy": [mean_wealthy],
        "mean_beautiful": [mean_beautiful],
        "mean_boring": [mean_boring],
        "mean_depressing": [mean_depressing],
        "median_safety": [median_safety],
        "median_lively": [median_lively],
        "median_wealthy": [median_wealthy],
        "median_beautiful": [median_beautiful],
        "median_boring": [median_boring],
        "median_depressing": [median_depressing],
        "std_safety": [std_safety],
        "std_lively": [std_lively],
        "std_wealthy": [std_wealthy],
        "std_beautiful": [std_beautiful],
        "std_boring": [std_boring],
        "std_depressing": [std_depressing],
    }), how='outer') 

ttl_df.to_csv('analysis/generate_stas/parks_perception_summary.csv', index=False)
```
